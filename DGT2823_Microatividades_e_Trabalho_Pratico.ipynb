## Parte 1: Microatividades Exploratórias (1 a 5)

### Microatividade 1: Leitura de Arquivo CSV com Pandas
**Objetivo:** Importar a biblioteca `pandas`, ler o conteúdo do arquivo CSV passando o separador de colunas (`sep=';'`), a engine (`engine='python'`) e a codificação (`encoding='utf-8'`), atribuir os dados a uma variável e exibi-los em tela.

```python
import pandas as pd
import numpy as np

# Caminho para o arquivo CSV do conjunto de dados
caminho_csv = 'conjunto_de_dados_trabalho_pratico_contextualizacao_atividade_pico_web.csv'

# Leitura do CSV com os parâmetros solicitados na Microatividade 1
df_original = pd.read_csv(
    caminho_csv,
    sep=';',
    engine='python',
    encoding='utf-8'
)

# Exibição dos dados lidos
print(df_original)
```

### Microatividade 2: Criação de Subconjunto de Dados
**Objetivo:** Criar uma nova variável contendo um subconjunto de dados com apenas 3 colunas do conjunto original (`ID`, `Date` e `Calories`) e exibir seu conteúdo em tela.

```python
# Criação do subconjunto com 3 colunas
df_subconjunto = df_original[['ID', 'Date', 'Calories']]

# Exibição do subconjunto de dados
print(df_subconjunto)
```

### Microatividade 3: Configuração de `max_rows` e Exibição com `to_string()`
**Objetivo:** Definir um novo valor para a propriedade `display.max_rows` (`9999`) nas opções de configuração do Pandas e imprimir o conjunto de dados original utilizando o método `.to_string()`.

```python
# Configuração do número máximo de linhas para 9999
pd.options.display.max_rows = 9999

# Exibição completa do conjunto de dados original como string
print(df_original.to_string())
```

### Microatividade 4: Exibição das Primeiras e Últimas 10 Linhas
**Objetivo:** Imprimir na tela apenas as primeiras 10 linhas (`.head(10)`) e as últimas 10 linhas (`.tail(10)`) do conjunto de dados original.

```python
print('--- Primeiras 10 Linhas ---')
print(df_original.head(10))

print('\n--- Últimas 10 Linhas ---')
print(df_original.tail(10))
```

**Objetivo:** Utilizar o método `.info()` para inspecionar a estrutura do DataFrame e responder às questões sobre total de linhas, total de colunas, dados nulos, tipos de dados e consumo de memória.

## Parte 2: Trabalho Prático — Tratamento e Limpeza de Dados (Passos 1 a 13)

### Passos 1 a 5: Importação e Verificação Inicial do Conjunto de Dados
**Objetivo:** Ler o arquivo CSV fornecido (composto pelas colunas `ID; Duration; Date; Pulse; Maxpulse; Calories`), atribuir a uma variável e verificar a importação imprimindo as informações gerais e as primeiras/últimas $N$ linhas.

```python
# Passos 1 a 4: Leitura do CSV
df_bruto = pd.read_csv(
    caminho_csv,
    sep=';',
    engine='python',
    encoding='utf-8'
)

# Passo 5a: Informações gerais
print('--- Passo 5a: Informações Gerais ---')
df_bruto.info()

# Passo 5b: Primeiras e últimas 5 linhas
print('\n--- Passo 5b: Primeiras 5 Linhas ---')
print(df_bruto.head(5))

print('\n--- Passo 5b: Últimas 5 Linhas ---')
print(df_bruto.tail(5))
```

### Passos 6 e 7: Cópia do Conjunto de Dados e Tratamento de Nulos em `Calories`
**Objetivo:** Criar uma cópia independente do DataFrame original (`.copy()`), substituir todos os valores nulos (`NaN`) da coluna `Calories` (linhas 18 e 28) por `0` utilizando `.fillna(0)` e imprimir o conjunto de dados para conferência.

```python
# Passo 6: Criação de uma cópia do conjunto de dados original
df_tratado = df_bruto.copy()

# Passo 7a: Substituição de valores nulos na coluna 'Calories' por 0
df_tratado['Calories'] = df_tratado['Calories'].fillna(0)

# Passo 7b: Impressão para verificar a substituição nas linhas 18 e 28
print(df_tratado.to_string())
```

### Passo 8: Substituição de Nulos em `Date` por `'1900/01/01'` e Primeira Tentativa com `to_datetime`
**Objetivo:** Substituir os valores nulos da coluna `Date` (linha 22) por `'1900/01/01'`, imprimir o DataFrame e tentar converter a coluna `Date` para o tipo `datetime` usando `pd.to_datetime()`.

> **Nota Pedagógica:** Como as datas das linhas 0 a 21 possuem aspas simples embutidas (ex.: `'2020/12/01'`), o Pandas infere o formato `'%Y/%m/%d'` (com aspas simples) e lança um `ValueError` ao encontrar `'1900/01/01'` (sem aspas simples) na posição 22 — exatamente como descrito no enunciado do Passo 9.

```python
# Passo 8a: Substituição dos valores nulos da coluna 'Date' por '1900/01/01'
df_tratado['Date'] = df_tratado['Date'].fillna('1900/01/01')

# Passo 8b: Impressão do conjunto de dados para conferência
print(df_tratado.to_string())

# Passo 8c: Transformação de 'Date' em datetime (capturando o erro esperado no Passo 9)
print('\n--- Tentativa de conversão com pd.to_datetime (Passo 8c) ---')
try:
    df_tratado['Date'] = pd.to_datetime(df_tratado['Date'])
except ValueError as e:
    print(f'ValueError esperado (Passo 9): {e}')
```

### Passo 9: Substituição de `'1900/01/01'` por `NaN` e Segunda Tentativa com `to_datetime`
**Objetivo:** Resolver o erro anterior substituindo `'1900/01/01'` por `NaN` (`np.nan`), repetir a tentativa de conversão com `pd.to_datetime()` e imprimir o conjunto de dados.

> **Nota Pedagógica:** Ao repetir o `pd.to_datetime()`, a linha 22 passa a ser ignorada como nula, mas o comando esbarra no segundo erro previsto no Passo 10: a string `'20201226'` (na linha 26) não corresponde ao formato `'%Y/%m/%d'`.

```python
# Passo 9a: Substituição do valor '1900/01/01' por NaN
df_tratado['Date'] = df_tratado['Date'].replace('1900/01/01', np.nan)

# Passo 9b: Repetição do passo de transformação para datetime (capturando o erro esperado no Passo 10)
print('--- Segunda tentativa de conversão com pd.to_datetime (Passo 9b) ---')
try:
    df_tratado['Date'] = pd.to_datetime(df_tratado['Date'])
except ValueError as e:
    print(f'ValueError esperado (Passo 10): {e}')

# Passo 9c: Impressão do conjunto de dados
print('\n--- Estado do DataFrame após o Passo 9 ---')
print(df_tratado.to_string())
```

### Passos 10 e 11: Correção do Registro `'20201226'` e Conversão Completa de `Date` para `datetime`
**Objetivo:**
1. **Passo 10:** Transformar especificamente o valor `'20201226'` (linha 26) combinando os métodos `.replace()` e `pd.to_datetime('20201226', format='%Y%m%d')`.
2. **Passo 11:** Executar novamente `pd.to_datetime()` sobre toda a coluna `Date` e imprimir o conjunto de dados atualizado.

```python
# Passo 10: Transformação específica do valor '20201226' combinando replace e to_datetime
df_tratado['Date'] = df_tratado['Date'].replace(
    '20201226',
    pd.to_datetime('20201226', format='%Y%m%d')
)

# Passo 11: Transformação de todos os dados da coluna 'Date' para o formato datetime
df_tratado['Date'] = pd.to_datetime(df_tratado['Date'])

# Impressão para verificar que todas as datas foram convertidas (e a linha 22 tornou-se NaT)
print(df_tratado.to_string())
```

### Passos 12 e 13: Remoção de Registros Nulos e Verificação Final
**Objetivo:**
1. **Passo 12:** Remover os registros contendo valores nulos (`.dropna(subset=['Date'])`), eliminando a linha 22 (`NaT`).
2. **Passo 13:** Imprimir o DataFrame final tratado (31 registros completos) e confirmar a estrutura final com `.info()`.

```python
# Passo 12: Remoção dos registros contendo valores nulos com base na coluna 'Date' (linha 22)
df_tratado = df_tratado.dropna(subset=['Date'])

# Passo 13: Impressão do DataFrame final e confirmação das transformações
print('--- Passo 13: DataFrame Final Tratado (31 Registros) ---')
print(df_tratado.to_string())

print('\n--- Informações Gerais do DataFrame Final ---')
df_tratado.info()
```
